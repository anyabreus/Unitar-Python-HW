## 1. Короткий опис завдання
Посилання на проблемний бриф (файл) і 2–3 речення: що рахуємо і для кого.

Рахуємо, який середній час завантаження сторінок сайту і позначаємо сторінки мітками швидкості ("швидко"/"нормально"/"повільно") для зручності. Це потрібно для фронтенд розробниці, щоб визначити, які сторінки потрібно оптимізувати і в якому порядку.

## 2. Дані
Список щонайменше з 8 записів-словників.

```python
records = [
    {"page": "home", "load_ms": 800, "images_optimized": True},
    {"page": "catalog", "load_ms": 1500, "images_optimized": False},
    {"page": "cart", "load_ms": 2200, "images_optimized": True},
    {"page": "login", "load_ms": 600, "images_optimized": True},
    {"page": "profile", "load_ms": 1000, "images_optimized": False},
    {"page": "checkout", "load_ms": 3500, "images_optimized": False},
    {"page": "search", "load_ms": 2000, "images_optimized": True},
    {"page": "about", "load_ms": 400, "images_optimized": True},
]
print(len(records))
```

## 3. Три показники
Спершу запишіть, що очікуєте (ручний розрахунок).

1. Загальна кількість проаналізованих сторінок: 8
2. Середній час завантаження сторінок: (800 + 1500 + 2200 + 600 + 1000 + 3500 + 2000 + 400) / 8 = 1500 ms
3. Частка сторінок з оптимізованими зображеннями: 5 / 8 * 100 = 62,5%

```python
pages_amount = len(records)
# закоментувала альтернативний код, який спершу написала,
# але потім у лекції побачила list comprehension, і захотілось спробувати :)
# total_load_ms = 0
total_load_ms = sum([page['load_ms'] for page in records])
# images_optimized_count = 0
images_optimized_count = len([page['page'] for page in records if page["images_optimized"]])
# for page in records:
#   total_load_ms += page['load_ms']
#   if page['images_optimized']:
#     images_optimized_count += 1

print("Загальна кількість проаналізованих сторінок:", pages_amount)
print("Середній час завантаження сторінок:", f"{round(total_load_ms/pages_amount)} ms")
print("Частка сторінок з оптимізованими зображеннями:",
      f"{(images_optimized_count/pages_amount):.1%}")
```

## 4. Класифікація (if / elif / else)
Які межі й чому?
- <= 1000 ms - "швидко"
- \> 2000 ms - "повільно"
- інше - "нормально"

```python
for page in records:
    if page['load_ms'] <= 1000:
      print("Сторінка", page['page'], "завантажується швидко")
    elif page['load_ms'] > 2000:
      print("Сторінка", page['page'], "завантажується повільно")
    else:
      print("Сторінка", page['page'], "завантажується нормально")
print("Бонус :) Найповільніші сторінки:",
      [page["page"] for page in records if page["load_ms"] > 2000])
```

## 5. Функції (щонайменше дві, з return)

```python
def classify_speed(page_speed):
    if page_speed <= 1000:
      return "швидко"
    elif page_speed > 2000:
      return "повільно"
    else:
      return "нормально"
chosen_page = records[2]
print("Сторінка", chosen_page['page'], "завантажується",
      classify_speed(chosen_page['load_ms']))

def get_slowest_pages(pages):
    return [page['page'] for page in pages
            if classify_speed(page['load_ms']) == 'повільно']
print("Найповільніші сторінки:", get_slowest_pages(records))

def get_img_optimized_share(pages):
    images_optimized_count = len([page['page'] for page in pages
                                  if page["images_optimized"]])
    return f"{(images_optimized_count / len(pages)):.1%}"
print("Частка сторінок з оптимізованими зображеннями:",
      get_img_optimized_share(records))

def get_avg_load_speed(pages):
    total_load_ms = sum([page['load_ms'] for page in pages])
    return f"{round(total_load_ms / len(pages))} ms"
print("Середній час завантаження сторінок:", get_avg_load_speed(records))
```

## 6. Перевірки (щонайменше три assert)

```python
assert classify_speed(1500) == 'нормально'
assert classify_speed(3000) == 'повільно'
assert get_avg_load_speed(records) == "1500 ms"
```
