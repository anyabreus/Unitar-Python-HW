### _Для двох числових стовпців обчисліть і запишіть: кількість, мінімум, максимум, розмах, середнє, медіану, моду, стандартне відхилення (std). Порахуйте кореляцію для однієї пари числових стовпців (не беріть стовпці-ідентифікатори). Одним-двома реченнями поясніть, що є «типовим», наскільки значення розкидані й чи пов’язані два стовпці (і чому це ще не доводить причину)._

```python
adl = ad_df['ADL']
sleep_quality = ad_df['SleepQuality']

print("ADL (Activities of Daily Living) Статистика:")
print(f"""Кількість: {adl.count()}
Мінімум: {adl.min():.2f}
Максимум: {adl.max():.2f}
Розмах: {(adl.max() - adl.min()):.2f}
Середнє: {adl.mean():.2f}
Медіана: {adl.median():.2f}
Мода: {adl.round(4).mode().tolist()[:5]}
Стандартне відхилення: {adl.std():.2f}""")

print("____________________________________________________________")

print("Sleep Quality Статистика:")
print(f"""Кількість: {sleep_quality.count()}
Мінімум: {sleep_quality.min():.2f}
Максимум: {sleep_quality.max():.2f}
Розмах: {(sleep_quality.max() - sleep_quality.min()):.2f}
Середнє: {sleep_quality.mean():.2f}
Медіана: {sleep_quality.median():.2f}
Мода: {sleep_quality.round(4).mode().tolist()[:5]}
Стандартне відхилення: {sleep_quality.std():.2f}""")
```

Тут в мене спочатку виникла SyntaxError при спробі вивести в print кожне значення з нового рядка через "\n", тому звернулась до АІ, і він підказав, що можна використати потрійні лапки.  
Також для зручності я округлила всі значення у виводі до 2 знаків після коми, але для моди я залишила 4 знаки після коми, щоб бачити точні значення, які повторюються найбільше разів, і обмежила вивід моди до 5 значень, щоб не перевантажувати.  
Загалом, і мінімальне, і максимальне значення співпадають з межами діапазону значень для цих стовпців, що наведено в описі до датасету, а середні значення і медіана показують, що вибірка досить збалансована, бо вони якраз знаходяться десь на середині тих діапазонів.

```python
print(ad_df['FunctionalAssessment'].corr(ad_df['Diagnosis']))
```

Значення `FunctionalAssessment` для цього датасету позначаються від 1 до 10, і чим нижче значення, тим менш функціональним є пацієнт. Кореляція між діагностованим Альцгеймером і Оцінкою Функціональності є помірною, і вказує на те, що в даній вибірці у пацієнтів з Альцгеймером може бути нижча функціональність.
