Метод IQR:

```python
q1 = df['CholesterolTotal'].quantile(0.25)
q3 = df['CholesterolTotal'].quantile(0.75)

iqr = q3 - q1

outliers = df[
    (df['CholesterolTotal'] < q1 - 1.5 * iqr) |
    (df['CholesterolTotal'] > q3 + 1.5 * iqr)
]

print(outliers)
```

### 3. Аналіз взаємозв'язків між ознаками

```python
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

plt.figure(figsize=(14, 10))
plt.imshow(corr, cmap='coolwarm', interpolation='nearest')

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

# for i in range(len(corr.columns)):
#     for j in range(len(corr.columns)):
#         plt.text(
#             j, i,
#             f'{corr.iloc[i, j]:.2f}',
#             ha='center',
#             va='center',
#             fontsize=7
#         )

plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
corr = df.corr(numeric_only=True)

# Отримуємо тільки верхню половину матриці
upper = corr.where(
    __import__('numpy').triu(
        __import__('numpy').ones(corr.shape), k=1
    ).astype(bool)
)

# Сортуємо кореляції за абсолютним значенням
strong_corr = upper.stack().sort_values(
    key=abs, ascending=False
)

print(strong_corr.head(10))
```

```python
grouped = df.groupby('Diagnosis')[
    ['Confusion', 'BehavioralProblems',
     'PersonalityChanges', 'Forgetfulness']
].mean()

print(grouped)
```

```python
grouped.plot(
    kind='bar',
    figsize=(10, 6),
    rot=0
)

plt.title('Comparison by Diagnosis')
plt.xlabel('Diagnosis (0 - No, 1 - Yes)')
plt.ylabel('Average value')
plt.legend(title='Indicators')
plt.tight_layout()
plt.show()
```
