# Мій перший ноутбук
Алгоритм із заняття: рахуємо суму покупок і перевіряємо безкоштовну доставку.

```python
prices = [250, 480, 199, 71]
total = 0
for price in prices:
    total = total + price
total
```

```python
if total >= 1000:
    print("Безкоштовна доставка")
else:
    print("Не вистачає до 1000:", 1000 - total)
```

**Завдання:** змініть список `prices` у першій клітинці так, щоб з'явилась «Безкоштовна доставка», і виконайте клітинки знову (Shift+Enter).
